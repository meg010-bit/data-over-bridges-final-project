```python
# Assigns each POGOH station into a neighborhood
import pandas as pd
pogoh = pd.read_csv('pogoh.tsv',sep='\t')
northOakland = []
centralOakland = []
westOakland = []
southOakland = []
shadyside = []
stripDistrict = []
downtown = []
southside = []
northside = []
squirrelHill = []
eastLiberty = []
wilkinsburg = []
lawrenceville = []
bloomfield = []
greenfield = []
hazelwood = []
hillDistrict = []
other = []
def assignNeighborhood(name):    
    name = name.lower()
    # North Oakland
    if any(x in name for x in ["n dithridge", "dithridge", "craig", "bayard", "filmore", "fifth ave", "n bellefield", "s bellefield", "centre & bellefield", "n oakland", "craig st"]):
        northOakland.append(name)

    # Central Oakland
    elif any(x in name for x in["forbes", "bouquet", "ohara", "schenley", "coltart", "university place", "sennott", "tcs hall", "atwood", "ruskin", "oakland ave"]):
        centralOakland.append(name)

    # South Oakland
    elif any(x in name for x in["bates", "zulema", "technology", "ward", "swinburne", "mackey", "dawson", "oakcliffe", "boquet", "s oakland"]):
        southOakland.append(name)

    # West Oakland
    elif any(x in name for x in["allequippa", "darragh", "parkview", "de soto", "terrence", "robinson", "mccormick", "west oakland"]):
        westOakland.append(name)
        
    # Southside
    elif any(x in name for x in ["sidney", "carson", "southside", "s 4th", "27th", "station square", "glasshouse"]):
        southside.append(name)

    # Shadyside
    elif any(x in name for x in ["walnut", "ellsworth", "ivy", "neville", "gross", "aiken", "summerlea", "rosetta", "centre & millvale"]):
        shadyside.append(name)

    # Squirrel Hill
    elif any(x in name for x in ["murray", "beacon", "wilkins ave", "forbes & murray"]):
        squirrelHill.append(name)

    # Strip District
    elif any(x in name for x in ["21st", "17th", "penn ave & 33rd", "penn ave & 34th", "7th", "strip district"]):
        stripDistrict.append(name)

    # Downtown
    elif any(x in name for x in ["market square", "stanwix", "first ave", "ross", "fourth ave", "penn ave & 10th"]):
        downtown.append(name)
        
    # East Liberty
    elif any(x in name for x in [
        "putnam", "bakery square", "negley", "pacific", "s pacific", "centre & negley", "tioga", "homewood", "hamilton", "bennett", "braddock", "n homewood"]):
        eastLiberty.append(name)

    # Northside
    elif any(x in name for x in ["federal", "ohio", "brighton", "allegheny station", "north shore", "w north ave"]):
        northside.append(name)
        
    # Lawrenceville
    elif any(x in name for x in["butler", "44th", "36th", "42nd", "friendship", "penn & 35th", "lawrenceville"]):
        lawrenceville.append(name)

    # Bloomfield
    elif any(x in name for x in [
        "liberty & gross", "liberty & millvale", "centre & millvale", "gross st", "bloomfield", "millvale", "friendship"]):
        bloomfield.append(name)
        
    # Wilkinsburg
    elif "wilkinsburg" in name or "park & ride" in name:
        wilkinsburg.append(name)
        
    # Greenfield
    elif any(x in name for x in [
        "swineburne", "lytle", "saline", "greenfield", "tecumseh", "eliza furnace", "second ave", "hazelwood", "murray hill", "bigelow"]):
        greenfield.append(name)

    # Hazelwood
    elif any(x in name for x in["tecumseh", "eliza furnace", "second ave", "hazelwood"]):
        hazelwood.append(name)
     
    elif "penn ave" in name:
        if any(x in name for x in ["10th", "stanwix", "market", "9th", "liberty"]):
            downtown.append(name)
        elif any(x in name for x in ["21st", "25th", "28th", "33rd", "35th", "40th"]):
            stripDistrict.append(name)
        elif any(x in name for x in ["whitfield", "putnam", "pacific", "negley"]):
            eastLiberty.append(name)
        elif any(x in name for x in ["gross", "millvale"]):
            bloomfield.append(name)
        else:
            centralOakland.append(name)
    
    elif "centre ave" in name:
        if any(x in name for x in ["addison", "heldman"]):
            hillDistrict.append(name)
        elif any(x in name for x in ["millvale", "gross", "aiken"]):
            bloomfield.append(name)
        elif any(x in name for x in ["negley", "bennett", "braddock", "tioga"]):
            eastLiberty.append(name)
        else:
            northOakland.append(name)
    else:
        other.append(name)

import pandas as pd
pogoh = pd.read_csv('pogoh.tsv',sep='\t')

pogoh['Name'].apply(assignNeighborhood)
neighborhoods = [
    ["North Oakland", northOakland],
    ["Central Oakland", centralOakland],
    ["South Oakland", southOakland],
    ["West Oakland", westOakland],
    ["Shadyside", shadyside],
    ["Southside", southside],
    ["Northside", northside],
    ["Squirrel Hill", squirrelHill],
    ["East Liberty", eastLiberty],
    ["Strip District", stripDistrict],
    ["Downtown", downtown],
    ["Wilkinsburg", wilkinsburg],
    ["Lawrenceville", lawrenceville],
    ["Bloomfield", bloomfield],
    ["Greenfield", greenfield],
    ["Hazelwood", hazelwood],
    ["Hill District", hillDistrict],
    ["Other", other]
]

maxCount = 0
bestNeighborhood = ""
for name, stations in neighborhoods:
    count = len(stations)
    if count > maxCount:
        maxCount = count
        bestNeighborhood = name

print(f"The neighborhood with the most POGOH stations is: {bestNeighborhood}")
print(f"Number of stations: {maxCount}")

print("\nStation counts by neighborhood:")
for name, stations in neighborhoods:
    print(f"{name}: {len(stations)}")

import matplotlib.pyplot as plt
pogoh['Name'].apply(assignNeighborhood)
hoodData = hood_data = {
    "North Oakland": northOakland,
    "Central Oakland": centralOakland,
    "South Oakland": southOakland,
    "West Oakland": westOakland,
    "Shadyside": shadyside,
    "Southside": southside,
    "Northside": northside,
    "Squirrel Hill": squirrelHill,
    "East Liberty": eastLiberty,
    "Strip District": stripDistrict,
    "Downtown": downtown,
    "Wilkinsburg": wilkinsburg,
    "Lawrenceville": lawrenceville,
    "Bloomfield": bloomfield,
    "Greenfield": greenfield,
    "Hazelwood": hazelwood,
    "Hill District": hillDistrict,
    "Other": other
}
station_counts = {hood: len(stations) for hood, stations in hood_data.items()}

# Convert to DataFrame for easier sorting/plotting
station_df = pd.DataFrame(list(station_counts.items()), columns=['Neighborhood', 'Number of Stations'])
station_df = station_df.sort_values('Number of Stations', ascending=False)

# Plot
plt.figure(figsize=(10,6))
plt.bar(station_df['Neighborhood'], station_df['Number of Stations'], color='blue', edgecolor='blue')
plt.title('Number of POGOH Stations per Neighborhood', fontsize=14)
plt.xlabel('Neighborhood', fontsize=12)
plt.ylabel('Number of Stations', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()    
```
