# City of Pittsburgh Trees
We chose to analyze the City of Pittsburgh Trees dataset to explore neighborhood sustainability through the lens of urban greenery. Trees are one of the most visible and impactful elements of a sustainable city as they improve air quality, offer shade, and reduce stormwater runoff. Our group aimed to understand the greenness of different Pittsburgh neighborhoods and whether trees are evenly distributed across the city or concentrated in specific areas. By doing this, we hoped to highlight both the neighborhoods investing in sustainability and those that may need more environmental attention.


# The Metric:
Our group’s main metric was sustainability, focusing on how neighborhoods support long-term environmental health. My sub-metric was biodiversity, specifically through the number and distribution of street trees. Trees help improve air quality, support local ecosystems, and signal a neighborhood’s investment in green infrastructure. By analyzing where trees are most and least concentrated, we aimed to assess which areas are thriving in biodiversity and which may need more environmental attention.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Suppress FutureWarnings from seaborn
warnings.simplefilter(action='ignore', category=FutureWarning)

# Load the city trees CSV
df = pd.read_csv("city_trees.csv", low_memory=False)

# Count trees per neighborhood
tree_counts = df['neighborhood'].value_counts()

# Plot the bar chart
plt.figure(figsize=(18,6))
sns.barplot(
    x=tree_counts.index,
    y=tree_counts.values,
    hue=tree_counts.index,    
    palette="Spectral",
    legend=False                   
)
plt.title("Tree Count by Neighborhood")
plt.ylabel("Tree Count")
plt.xlabel("Neighborhood")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("city_trees.csv", low_memory=False)
top10 = df['neighborhood'].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top10.index, y=top10.values, palette="Spectral")
plt.title("Top 10 Neighborhoods by Tree Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
